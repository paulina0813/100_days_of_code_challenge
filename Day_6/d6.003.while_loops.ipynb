# For vs While Loops

- For loops are great for when you want to iterate over something and you need to do something with each thing that you're iterating over
- While loops are great when you don't really care what number in a sequence you're in, which item you're iterating through in a list and you simply want to carry out some sort of functionality many, many times until some sort of condition that you set
- In for loops you are setting ahead of time how many times the loop is going to run through
- A while loop will continue running as long as the condition is being met

## Infinite loops

- When you have a while loop and its condition never becomes False, then its known as an infinite loop.

For example:

while 5 > 3:
    #Do this
    #Then do this
    #Then do this

This is an infinite loop because 5 is ALWAYS going to be greater than 3, which means that the code will run indefenitely
