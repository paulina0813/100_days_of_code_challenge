# Create your own function

- First you gotta start with the keyword "def" because we're creating or defining our own function
- Enter the function name and put the parenthesis
- What differentiates a function from a variable are the parenthesis
- Add a colon
- Everything that comes after this line and is indented belongs to the function

```python
def my_function():
    print("Hello")
    print("Bye")
```

With the code above, you have defined the function but if you run the code, nothing will print. This is because you have yet to execute the function. You can write different functions ahead of time and then you trigger (call the function) them when you actually need them. 

To call the function, you need to write the name of the function and add any necessary arguments. In this case, the function doesn't require any arguments so the parenthesis can be left blank again.

```python
my_function()
```
