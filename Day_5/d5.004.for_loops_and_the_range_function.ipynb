By default, the range function will step through all the numbers from the start to the end, and increase by 1. If you want to increase by any other number, then you have to add another comma and specify how large you want the step to be.

for number in range(a,b, c):
    print(number)

Where c is the step

```python
#For loop with range
for number in range(1, 11, 3):
    print(number)
```

How do we add all the numbers from 1 to 100?

```python
total = 0
for number in range(1.101):
    total += number
print(total)
```
